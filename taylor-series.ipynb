## Taylor series

The Taylor series is a mathematical method used to represent complex, infinitely differentiable functions as simpler power series, effectively approximating them as polynomials. 

To understand why this is useful, consider that complicated functions like trigonometric (e.g., sine, cosine) or exponential functions are mathematically difficult to compute directly. Polynomials, on the other hand, are relatively easy to calculate and differentiate. By using Taylor series to approximate these complex functions as polynomials, we can vastly simplify calculations. In fact, this is the exact method pocket calculators use to efficiently compute sine, cosine, or exponential values.

### How the Formula is Built

The goal is to represent a function $f(x)$ as a power series: 
$P(x) = a_0 + a_1x + a_2x^2 + a_3x^3 + \dots$.

To figure out what the coefficients ($a_0, a_1, a_2$, etc.) should be, the Taylor series relies on taking successive derivatives of the original function at a specific "expansion point," typically chosen as $x_0 = 0$. 

If you take the derivatives of a polynomial power series and set $x=0$, a distinct pattern emerges: all the terms containing $x$ drop out, and the $k$-th derivative of the function evaluated at $0$, or $f^{(k)}(0)$, is equal to $k! \cdot a_k$ (where $k!$ represents "k factorial", e.g., $4! = 4 \cdot 3 \cdot 2 \cdot 1$). 

By solving this relationship for the coefficient $a_k$, we get:
$a_k = \frac{f^{(k)}(0)}{k!}$.

### The Taylor Series Formula

Substituting that coefficient back into the power series gives us the formal Taylor series definition for a function expanded around $x_0 = 0$:

$T_f(x) := \sum_{k=0}^{\infty} \frac{f^{(k)}(0)}{k!} x^k$.

While it is common to expand around $0$, the formula can be generalized to approximate the function around **any specific point $x_0$**:

$\sum_{k=0}^{\infty} \frac{f^{(k)}(x_0)}{k!} (x - x_0)^k$.

### Taylor Polynomials and Accuracy

In most practical applications, the infinite Taylor series converges to precisely equal the original function. However, because we cannot compute an infinite number of terms, we use partial sums of this series, which are called **Taylor polynomials of the $n$-th degree**. 

A critical rule of Taylor polynomials is that **the approximation is most accurate in a small neighborhood around the expansion point**. The further you move the value of $x$ away from that expansion point, the less accurate the polynomial approximation becomes. 

### A Quick Example
Consider the exponential function $f(x) = e^x$. It is infinitely differentiable, and its derivative is always $e^x$. If we expand around $x=0$, every single derivative evaluated at $0$ equals $1$ (since $e^0 = 1$). Plugging this into the formula gives the series $\sum_{k=0}^{\infty} \frac{1}{k!} x^k$, an approximation so reliable that it is frequently used to mathematically define the exponential function itself.
