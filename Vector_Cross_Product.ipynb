# Cross Product Properties

Let vectors $\overrightarrow{u}$ and $\overrightarrow{v}$:

$$\overrightarrow{u}= \{a_1 \overrightarrow{i} + a_2\overrightarrow{j} + a_3\overrightarrow{k}  \}$$

$$\overrightarrow{v}= \{b_1 \overrightarrow{i} + b_2\overrightarrow{j} + b_3\overrightarrow{k}  \}$$

## Cross Product:
$$ \left(a_{2}b_3\overrightarrow{i} - a_{3}b_{2}\overrightarrow{i}\right) - \left(a_{1}b_3\overrightarrow{j} - a_{3}b_{1}\overrightarrow{j}\right) + \left(a_{1}b_2\overrightarrow{k} - a_{1}b_{2}\overrightarrow{k}\right) $$

For nonzero vectors in $ℝ^3$, the cross product of $\overrightarrow{v}$ and $\overrightarrow{u}$ is a vector that is $\mathbf{perpendicular}$ to both of the given vectors.

Properties: 

1. $\overrightarrow{u} \times \overrightarrow{v} = - \left(\overrightarrow{v} \times \overrightarrow{u} \right)$.
2. $\overrightarrow{u} \times \left(\overrightarrow{v} +\overrightarrow{w}\right) = \overrightarrow{u} \times \overrightarrow{v} + \overrightarrow{u} \times \overrightarrow{w} $.
3. $\overrightarrow{u} \times \overrightarrow{u} = 0$.
4. $\overrightarrow{u}.\left(\overrightarrow{v} \times \overrightarrow{w}\right) = \left(\overrightarrow{u} \times \overrightarrow{v}\right) . \overrightarrow{w}$
5. The cross product is not associative: $\overrightarrow{u} \times \left(\overrightarrow{v} \times \overrightarrow{w}\right) \ne \left(\overrightarrow{u} \times \overrightarrow{v}\right) \times \overrightarrow{w} $


# Geometric Properties:

1. if $\overrightarrow{u} = k\overrightarrow{v}$ then $\overrightarrow{u} \times \overrightarrow{v}=0 $ (where k is scaler and $\overrightarrow{v}$ can be expressed as a scaler product of $\overrightarrow{u}$)
2. The resulting vector $\overrightarrow{u} \times \overrightarrow{v}$ will be orthogonal to both $\overrightarrow{u}$ and $\overrightarrow{v}$

Magnitude $\left|\overrightarrow{u} \times \overrightarrow{v} \right| = \left|\overrightarrow{u}\right|\left|\overrightarrow{v}\right|sin\theta$, which is equal to the area of parallelogram with adjacent sides $\overrightarrow{u}$ and $\overrightarrow{v}$ 

Height of parallelogram:

$$h=\left|\overrightarrow{v}\right|sin\theta$$

Area: $$h\left|\overrightarrow{u}\right| = \left|\overrightarrow{v}\right| \left|\overrightarrow{u}\right|sin\theta$$

Magnitude:

$$\left|\overrightarrow{u} \times \overrightarrow{v}\right| = \sqrt{a^2+b^2+c^2}$$

Unit Vector orthogonal to both vectors:

$$\frac{\overrightarrow{u} \times \overrightarrow{v}}{\left|\overrightarrow{u} \times \overrightarrow{v}\right|}$$
